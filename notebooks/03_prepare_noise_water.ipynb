```python
from pathlib import Path
import pandas as pd

base = Path("../data/raw/monitoring_2026-05-21_2026-06-20")

noise_files = list(base.rglob("*Zaj.xlsx"))
water_files = list(base.rglob("*Felszin_alatti_viz.xlsx"))

print(f"Noise files: {len(noise_files)}")
print(f"Water files: {len(water_files)}")
```

```python
from pathlib import Path
import pandas as pd

base = Path("../data/raw/monitoring_2026-05-21_2026-06-19")

noise_files = list(base.rglob("*_Zaj.xlsx"))
water_files = list(base.rglob("*_Felszin_alatti_viz.xlsx"))

print(f"Noise files: {len(noise_files)}")
print(f"Water files: {len(water_files)}")
```

```python
noise_df = pd.read_excel(noise_files[0])
water_df = pd.read_excel(water_files[0])

print("Noise columns:")
print(noise_df.columns.tolist())

print("\nWater columns:")
print(water_df.columns.tolist())

display(noise_df.head())

display(water_df.head())
```

```python
noise_all = pd.concat(
    [pd.read_excel(f) for f in noise_files],
    ignore_index=True,
)

print("Noise records:", len(noise_all))
print("Stations:", noise_all["Location"].nunique())

display(noise_all.head())
```

```python
water_all = pd.concat(
    [pd.read_excel(f) for f in water_files],
    ignore_index=True,
)

print("Water records:", len(water_all))
print("Stations:", water_all["Location"].nunique())

display(water_all.head())
```

```python
noise_clean = noise_all.rename(
    columns={
        "Mérőeszköz": "measurement_type",
        "érték": "value",
        "mértékegység": "unit",
        "Location": "location",
    }
)

noise_clean["timestamp"] = pd.to_datetime(
    noise_clean["timestamp"],
    format="%Y-%m-%d-%H-%M",
)

noise_clean["value"] = pd.to_numeric(
    noise_clean["value"],
    errors="coerce",
)

print(noise_clean.dtypes)

display(noise_clean.head())
```

```python
noise_clean["measurement_type"].value_counts()
```

```python
noise_clean.groupby("measurement_type")["value"].describe()
```

```python
water_clean = water_all.rename(
    columns={
        "Mérőeszköz": "measurement_type",
        "érték": "value",
        "mértékegység": "unit",
        "Location": "location",
    }
)

water_clean["timestamp"] = pd.to_datetime(
    water_clean["timestamp"],
    format="%Y-%m-%d-%H-%M",
)

water_clean["value"] = pd.to_numeric(
    water_clean["value"],
    errors="coerce",
)

print(water_clean["measurement_type"].value_counts())

display(
    water_clean.groupby("measurement_type")["value"].describe()
)
```

```python
air_locations = sorted(
    pd.read_csv(
        "../data/processed/air_measurements_cleaned.csv"
    )["location"].dropna().unique()
)

noise_locations = sorted(
    noise_clean["location"].dropna().unique()
)

water_locations = sorted(
    water_clean["location"].dropna().unique()
)

locations = pd.DataFrame({
    "Air": pd.Series(air_locations),
    "Noise": pd.Series(noise_locations),
    "Water": pd.Series(water_locations),
})

display(locations)
```

```python
noise_clean.to_csv("../data/processed/noise_measurements_cleaned.csv", index=False)

water_clean.to_csv("../data/processed/water_measurements_cleaned.csv", index=False)
```
