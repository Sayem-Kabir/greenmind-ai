```python
from pathlib import Path
import pandas as pd
```

```python
# DKV data location
dkv_path = Path("../data/raw/dkv")

# Show everything inside the DKV folder
for item in sorted(dkv_path.iterdir()):
    print(item.name)
```

```python
stops = pd.read_excel(dkv_path / "List of bus stops.xlsx")

print(stops.shape)
display(stops.head())
```

```python
summary_path = dkv_path / "Summary stop statistics"

for file in sorted(summary_path.iterdir()):
    print(file.name)
```

```python
year_2026 = summary_path / "2026"

for file in sorted(year_2026.iterdir()):
    print(file.name)
```

```python
may_stats = pd.read_excel(year_2026 / "05.xlsx")

print(may_stats.shape)
display(may_stats.head())
```

```python
may_stats = pd.read_excel(
    year_2026 / "05.xlsx",
    header=4
)

print(may_stats.shape)
print(may_stats.columns.tolist())

display(may_stats.head())
```

```python
for i, col in enumerate(may_stats.iloc[2]):
    print(f"{i}: {col}")
```

```python
# Show the three header rows together for every column
for i in range(may_stats.shape[1]):
    level_1 = may_stats.iloc[0, i]
    level_2 = may_stats.iloc[1, i]
    level_3 = may_stats.iloc[2, i]

    print(f"{i}: {level_1} | {level_2} | {level_3}")
```

```python
clean_stats = may_stats.iloc[3:].copy()

clean_stats.columns = [
    "stop_id",
    "stop_name",
    "planned_stops_apc",
    "planned_stops_all",
    "passengers_in_total",
    "passengers_in_avg",
    "passengers_out_total",
    "passengers_out_avg",
    "passenger_frequency_total",
    "passenger_frequency_avg",
    "occupancy_arrival_total",
    "occupancy_arrival_avg",
    "occupancy_departure_total",
    "occupancy_departure_avg",
    "delay_avg",
]

clean_stats.reset_index(drop=True, inplace=True)

print(clean_stats.shape)
display(clean_stats.head())
```

```python
numeric_columns = [
    "planned_stops_apc",
    "planned_stops_all",
    "passengers_in_total",
    "passengers_in_avg",
    "passengers_out_total",
    "passengers_out_avg",
    "passenger_frequency_total",
    "passenger_frequency_avg",
    "occupancy_arrival_total",
    "occupancy_arrival_avg",
    "occupancy_departure_total",
    "occupancy_departure_avg",
    "delay_avg",
]

for column in numeric_columns:
    clean_stats[column] = pd.to_numeric(clean_stats[column], errors="coerce")

clean_stats[numeric_columns].info()
```

```python
top_stops = (
    clean_stats[
        [
            "stop_name",
            "passenger_frequency_total",
            "passengers_in_total",
            "passengers_out_total",
            "occupancy_arrival_total",
            "occupancy_departure_total",
        ]
    ]
    .sort_values("passenger_frequency_total", ascending=False)
)

display(top_stops.head(20))
```

```python
# Remove the summary row
clean_stats = clean_stats[clean_stats["stop_name"] != "Total"].copy()

# Replace invalid passenger counts (-1) with 0
clean_stats["passengers_in_total"] = (
    clean_stats["passengers_in_total"]
    .replace(-1, 0)
)

print(clean_stats.shape)
```

```python
metrics = [
    "passenger_frequency_total",
    "passengers_in_total",
    "passengers_out_total",
    "occupancy_arrival_total",
    "occupancy_departure_total",
]

clean_stats[metrics].describe().T
```

```python
from sklearn.preprocessing import MinMaxScaler

score_features = [
    "passenger_frequency_total",
    "passengers_in_total",
    "passengers_out_total",
    "occupancy_arrival_total",
    "occupancy_departure_total",
]

scaler = MinMaxScaler()

normalized = scaler.fit_transform(
    clean_stats[score_features].fillna(0)
)

normalized = pd.DataFrame(
    normalized,
    columns=[f"{c}_norm" for c in score_features]
)

clean_stats = pd.concat(
    [clean_stats.reset_index(drop=True), normalized],
    axis=1,
)

display(clean_stats.head())
```

```python
# Average occupancy
clean_stats["occupancy_avg"] = (
    clean_stats["occupancy_arrival_total_norm"] +
    clean_stats["occupancy_departure_total_norm"]
) / 2

# Traffic Activity Score (0–100)
clean_stats["traffic_activity_score"] = (
    0.40 * clean_stats["passenger_frequency_total_norm"] +
    0.25 * clean_stats["passengers_in_total_norm"] +
    0.25 * clean_stats["passengers_out_total_norm"] +
    0.10 * clean_stats["occupancy_avg"]
) * 100

clean_stats[
    [
        "stop_name",
        "traffic_activity_score"
    ]
].sort_values(
    "traffic_activity_score",
    ascending=False
).head(20)
```

```python
traffic_score = (
    clean_stats
    .groupby("stop_name", as_index=False)
    .agg(
        traffic_activity_score=("traffic_activity_score", "mean"),
        passenger_frequency_total=("passenger_frequency_total", "sum"),
        passengers_in_total=("passengers_in_total", "sum"),
        passengers_out_total=("passengers_out_total", "sum"),
    )
    .sort_values(
        "traffic_activity_score",
        ascending=False
    )
)

display(traffic_score.head(20))
```

```python
# Keep only the columns we need
stop_locations = stops[["Bus stop", "Coordinates"]].copy()

# Rename to match the traffic_score dataframe
stop_locations = stop_locations.rename(
    columns={
        "Bus stop": "stop_name",
        "Coordinates": "coordinates",
    }
)

# Merge
traffic_locations = traffic_score.merge(
    stop_locations,
    on="stop_name",
    how="left"
)

print(traffic_locations.shape)
display(traffic_locations.head(20))
```

```python
# Split "longitude, latitude" into separate numeric columns
traffic_locations[["longitude", "latitude"]] = (
    traffic_locations["coordinates"]
    .str.split(",", expand=True)
)

traffic_locations["longitude"] = pd.to_numeric(
    traffic_locations["longitude"].str.strip(),
    errors="coerce",
)

traffic_locations["latitude"] = pd.to_numeric(
    traffic_locations["latitude"].str.strip(),
    errors="coerce",
)

# Create one representative location per physical stop
traffic_locations_final = (
    traffic_locations
    .groupby("stop_name", as_index=False)
    .agg(
        longitude=("longitude", "mean"),
        latitude=("latitude", "mean"),
        traffic_activity_score=("traffic_activity_score", "first"),
        passenger_frequency_total=("passenger_frequency_total", "first"),
        passengers_in_total=("passengers_in_total", "first"),
        passengers_out_total=("passengers_out_total", "first"),
    )
)

print(traffic_locations_final.shape)
display(traffic_locations_final.head(20))
```

```python
from pathlib import Path

output_dir = Path("../data/processed")
output_dir.mkdir(exist_ok=True)

traffic_locations_final.to_csv(
    output_dir / "traffic_activity.csv",
    index=False
)

print("Saved:", output_dir / "traffic_activity.csv")
```
